## **Workflow Example:** Combining Libraries

### 1. Import Required Libraries

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

### 2. Load the Dataset

```python
flights = sns.load_dataset('flights')
flights.head()
```

<!-- ### 3. Data Exploration -->

```python
# Basic information and statistics
# print(flights.info())
# print(flights.describe())

# # Check for missing values
# print(flights.isnull().sum())

# # Unique years and months
# print(f"Years: {flights['year'].unique()}")
print(f"Months: {flights['month'].unique()}")
```

<!-- ### 4. Statistical Summary with NumPy -->

```python
# Calculate monthly passenger mean and standard deviation
monthly_passengers = flights['passengers']
mean_passengers = np.mean(monthly_passengers)
std_passengers = np.std(monthly_passengers)

print(f"Mean Passengers: {mean_passengers}")
print(f"Standard Deviation Passengers: {std_passengers}")
```

### 5. Visualization with Seaborn

**Line Plot Passenger Over Time**

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=flights, x='year', y='passengers', hue='month', marker='o')
plt.title('Number of Passengers Over Time by Month', fontsize=18, fontweight="bold", color="navy")
plt.xlabel('Year', fontweight="bold", color="navy")
plt.ylabel('Number of Passengers', fontweight="bold", color="navy")
plt.legend(title='Month', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
